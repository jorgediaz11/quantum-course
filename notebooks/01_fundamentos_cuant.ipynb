## De Bits a Qubits — Visualización y Manipulación

**Objetivo:** Comprender la diferencia entre bits clásicos y qubits, dominar la notación de Dirac, 
entender el proceso de medición y visualizar estados en la Esfera de Bloch.

**Herramientas:** Qiskit, NumPy, Matplotlib

```python
# Imports para toda la Fase 2
import numpy as np
import matplotlib.pyplot as plt
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_bloch_multivector, plot_histogram

# Configuración de visualización
plt.style.use('seaborn-v0_8-darkgrid')
print("✅ Librerías cargadas correctamente")
```

```python
# ===================================================
# TAREA 2.1 — Bits clásicos vs Qubits (experimento)
# ===================================================
from qiskit.quantum_info import Statevector

# --- PARTE A: 3 BITS CLÁSICOS ---
print("=" * 55)
print("PARTE A: 3 BITS CLÁSICOS")
print("=" * 55)

bits = [1, 0, 1]
print(f"Estado actual:      {bits}")
print(f"Combinaciones:      2^3 = {2**3}")
print("→ El sistema ocupa UNA sola de esas 8 combinaciones a la vez.")

# --- PARTE B: 3 QUBITS ---
print("\n" + "=" * 55)
print("PARTE B: 3 QUBITS")
print("=" * 55)

# Paso 1: creamos un circuito con 3 qubits (inician en |000>)
qc = QuantumCircuit(3)
print("Estado inicial:     |000>")

# Paso 2: Aplicamos Hadamard (H) a los tres qubits → superposición
qc.h([0, 1, 2])

# Paso 3: Inspeccionamos el vector de estados (SIN medir, solo en simulador)
sv = Statevector.from_instruction(qc)
probs = sv.probabilities_dict()

print(f"Estados simultáneos: {len(sv.data)}")
print("\nProbabilidad de cada estado base:")

# Mostramos las probabilidades ordenadas
for estado, p in sorted(probs.items()):
    print(f"   |{estado}>  →  {p:.4f}   ({p*100:5.2f}%)")

# Paso 4: Verificamos que la suma de probabilidades sea 1 (100%)
print(f"\nSuma de probabilidades: {sum(probs.values()):.6f}  ✓")

# Mostramos el circuito visualmente
print("\nCircuito construido:")
print(qc.draw("text"))
```

```python
# ===================================================
# TAREA 2.2 — Notación de Dirac (Bra-Ket) con NumPy
# ===================================================
import numpy as np

print("=" * 60)
print("NOTACIÓN DE DIRAC — REPRESENTACIÓN VECTORIAL")
print("=" * 60)

# Paso 1: Definir los estados base como vectores columna (Kets)
ket_0 = np.array([1, 0])
ket_1 = np.array([0, 1])

print("\n📐 Estados base (Kets):")
print(f"|0⟩ = {ket_0}")
print(f"|1⟩ = {ket_1}")

# Paso 2: Construir el estado de superposición |+⟩
# |+⟩ = (1/√2)|0⟩ + (1/√2)|1⟩
alpha = 1 / np.sqrt(2)
beta = 1 / np.sqrt(2)
ket_plus = alpha * ket_0 + beta * ket_1

print(f"\n🌊 Estado de superposición |+⟩:")
print(f"|+⟩ = ({alpha:.4f})|0⟩ + ({beta:.4f})|1⟩")
print(f"|+⟩ = {np.round(ket_plus, 4)}")

# Paso 3: Verificar la normalización (La suma de probabilidades debe ser 1)
# La probabilidad es el valor absoluto al cuadrado de la amplitud: |α|² + |β|²
prob_0 = np.abs(alpha)**2
prob_1 = np.abs(beta)**2
total_prob = prob_0 + prob_1

print(f"\n✓ Verificación de normalización (Regla de Born):")
print(f"  P(0) = |α|² = {prob_0:.4f}")
print(f"  P(1) = |β|² = {prob_1:.4f}")
print(f"  Suma total = {total_prob:.4f}  (Debe ser 1.0) ✓")

# Paso 4: Demostrar ortogonalidad (Producto interno o "Bra-Ket")
# <0|1> = ket_0† · ket_1 (como son reales, solo es el producto punto)
inner_product = np.vdot(ket_0, ket_1)

print(f"\n✓ Verificación de ortogonalidad:")
print(f"  ⟨0|1⟩ = {inner_product}")
print(f"  → Al ser 0, los estados son ortogonales (mutuamente excluyentes).")
```

```python
# ===================================================
# TAREA 2.3 — Medición y Colapso (Regla de Born)
# ===================================================
print("=" * 60)
print("MEDICIÓN Y COLAPSO — REGLA DE BORN")
print("=" * 60)

# 1. Creamos un circuito con 1 qubit y 1 bit clásico
qc = QuantumCircuit(1, 1)

# 2. Aplicamos Hadamard para crear superposición |+⟩
qc.h(0)
print("\n🌊 Estado antes de medir: |+⟩ (superposición 50/50)")
print(qc.draw("text"))

# 3. Medimos el qubit 0 y guardamos en el bit clásico 0
qc.measure(0, 0)
print("\n📏 Estado después de medir: El qubit colapsa a |0⟩ o |1⟩")
print(qc.draw("text"))

# 4. Simulamos el proceso 1000 veces (shots)
simulator = AerSimulator()
result = simulator.run(qc, shots=1000).result()
counts = result.get_counts()

# 5. Mostramos los resultados
print(f"\n📊 Resultados de 1000 mediciones: {counts}")
prob_0 = counts.get('0', 0) / 1000 * 100
prob_1 = counts.get('1', 0) / 1000 * 100
print(f"   → Probabilidad de '0': {prob_0:.1f}%")
print(f"   → Probabilidad de '1': {prob_1:.1f}%")

# 6. Visualización del histograma
plot_histogram(counts, title="Distribución de mediciones (Colapso de la función de onda)")
plt.show()

print("\n💡 Nota clave: Una vez medido, la superposición se destruye irreversiblemente.")
```

```python
# ===================================================
# TAREA 2.4 — Esfera de Bloch (Visualización Geométrica)
# ===================================================
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_multivector
import matplotlib.pyplot as plt

print("=" * 60)
print("ESFERA DE BLOCH — REPRESENTACIÓN GEOMÉTRICA")
print("=" * 60)

# 1. Estado |0⟩ (Polo Norte)
print("\n🔵 1. Estado |0⟩ (Estado inicial por defecto)")
qc_0 = QuantumCircuit(1)
sv_0 = Statevector.from_instruction(qc_0)
fig_0 = plot_bloch_multivector(sv_0)
plt.suptitle("Estado |0⟩ → Polo Norte (Z+)", fontsize=14, y=1.05)
plt.show()

# 2. Estado |+⟩ (Superposición en el Ecuador)
print("\n🟣 2. Estado |+⟩ (Tras aplicar Hadamard)")
qc_plus = QuantumCircuit(1)
qc_plus.h(0)  # Rotación de |0⟩ al eje X+
sv_plus = Statevector.from_instruction(qc_plus)
fig_plus = plot_bloch_multivector(sv_plus)
plt.suptitle("Estado |+⟩ = H|0⟩ → Ecuador (Eje X+)", fontsize=14, y=1.05)
plt.show()

# 3. Estado |1⟩ (Polo Sur)
print("\n🔴 3. Estado |1⟩ (Tras aplicar Pauli-X)")
qc_1 = QuantumCircuit(1)
qc_1.x(0)  # Rotación de |0⟩ al eje Z-
sv_1 = Statevector.from_instruction(qc_1)
fig_1 = plot_bloch_multivector(sv_1)
plt.suptitle("Estado |1⟩ = X|0⟩ → Polo Sur (Z-)", fontsize=14, y=1.05)
plt.show()

print("\n💡 Nota clave: La compuerta H no es un 'generador de aleatoriedad',")
print("   es una ROTACIÓN de 180° alrededor del eje diagonal X+Z en la esfera.")
```

```python
# ===================================================
# TAREA 2.5 — Primeras Compuertas y No Conmutatividad
# ===================================================
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_multivector
import matplotlib.pyplot as plt

print("=" * 65)
print("COMPUERTAS CUÁNTICAS Y NO CONMUTATIVIDAD")
print("=" * 65)

# --- PARTE A: Efecto individual de las compuertas ---
print("\n🔹 PARTE A: Rotaciones básicas desde el estado |0⟩")

compuertas = {
    'X (NOT)': lambda qc: qc.x(0),
    'Y': lambda qc: qc.y(0),
    'Z': lambda qc: qc.z(0),
    'H (Hadamard)': lambda qc: qc.h(0)
}

for nombre, aplicar in compuertas.items():
    qc = QuantumCircuit(1)
    aplicar(qc)
    sv = Statevector.from_instruction(qc)
    
    print(f"\n[{nombre}]")
    print(qc.draw("text"))
    
    # Visualizar en Bloch
    fig = plot_bloch_multivector(sv)
    plt.suptitle(f"Estado tras aplicar {nombre} a |0⟩", fontsize=12, y=1.05)
    plt.show()

# --- PARTE B: Demostración de No Conmutatividad (El orden SÍ importa) ---
print("\n" + "=" * 65)
print("🔹 PARTE B: NO CONMUTATIVIDAD (El orden altera el resultado)")
print("=" * 65)

# Caso 1: Primero X, luego H  (H · X)
qc_1 = QuantumCircuit(1)
qc_1.x(0)
qc_1.h(0)
sv_1 = Statevector.from_instruction(qc_1)

print("\n1. Circuito: |0⟩ → X → H")
print(qc_1.draw("text"))
print(f"Vector de estado resultante: {np.round(sv_1.data, 3)}")

fig_1 = plot_bloch_multivector(sv_1)
plt.suptitle("Orden: X luego H", fontsize=12, y=1.05)
plt.show()

# Caso 2: Primero H, luego X  (X · H)
qc_2 = QuantumCircuit(1)
qc_2.h(0)
qc_2.x(0)
sv_2 = Statevector.from_instruction(qc_2)

print("\n2. Circuito: |0⟩ → H → X")
print(qc_2.draw("text"))
print(f"Vector de estado resultante: {np.round(sv_2.data, 3)}")

fig_2 = plot_bloch_multivector(sv_2)
plt.suptitle("Orden: H luego X", fontsize=12, y=1.05)
plt.show()

# Verificación matemática
print("\n💡 CONCLUSIÓN:")
print(f"¿Son iguales los estados finales? {np.allclose(sv_1.data, sv_2.data)}")
print("→ En computación cuántica, A·B ≠ B·A. ¡El orden de las operaciones es crucial!")
```

## 📚 RESUMEN DE LA FASE 2: FUNDAMENTOS Y EXPLORACIÓN CUÁNTICA

### 🎯 Conceptos Clave Dominados

#### 1. **Bits vs Qubits** (Tarea 2.1)
- **Bit clásico**: Solo puede estar en 0 o 1 (un estado a la vez)
- **Qubit**: Puede estar en superposición α|0⟩ + β|1⟩ (múltiples estados simultáneamente)
- **Escalabilidad**: 3 bits = 1 estado | 3 qubits = 8 estados simultáneos

#### 2. **Notación de Dirac** (Tarea 2.2)
- **Ket** |ψ⟩: Vector columna que representa el estado cuántico
- **Bra** ⟨ψ|: Vector fila conjugado
- **Producto interno** ⟨φ|ψ⟩: Mide la "similitud" entre estados
- **Normalización**: |α|² + |β|² = 1 (suma de probabilidades = 100%)

#### 3. **Medición y Colapso** (Tarea 2.3)
- **Regla de Born**: P(0) = |α|², P(1) = |β|²
- **Colapso**: Al medir, la superposición se destruye irreversiblemente
- **Post-medición**: El qubit queda en el estado medido (0 o 1)

#### 4. **Esfera de Bloch** (Tarea 2.4)
- **Representación geométrica**: Cualquier estado puro de un qubit es un punto en la esfera
- **Polo Norte**: Estado |0⟩
- **Polo Sur**: Estado |1⟩
- **Ecuador**: Estados de superposición (como |+⟩)

#### 5. **Compuertas Cuánticas** (Tarea 2.5)
- **X (NOT)**: Rotación 180° alrededor del eje X (|0⟩ → |1⟩)
- **Y**: Rotación 180° alrededor del eje Y
- **Z**: Rotación 180° alrededor del eje Z (cambia la fase)
- **H (Hadamard)**: Crea superposición (|0⟩ → |+⟩)
- **No conmutatividad**: El orden importa (H·X ≠ X·H)
